```python
import pandas as pd
```

```python
df=pd.read_csv("titanic.csv",usecols=["Age","Fare"])
```

```python
df
```

```python
df.isnull().sum()
```

```python
# no of missing values in age
n=df["Age"].isnull().sum()
```

```python
n
```

```python
# create a copy
df["Age_imputed"]=df["Age"]
```

```python
# Randomly sample from existing values
random_values =df['Age'].dropna().sample(
    n=df['Age'].isnull().sum(),
    random_state=42
).values


#filling
df.loc[
    df['Age_imputed'].isnull(),
    'Age_imputed'
] = random_values
```

```python
df.sample(n=10)
```
